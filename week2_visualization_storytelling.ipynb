# Week 2 – Advanced Data Visualization and Storytelling with Python
## The Development Story: Income, Health and Population

This notebook develops a visual narrative using the public Gapminder dataset. The focus is not only on making charts, but on choosing visuals that answer a sequence of questions for a non-technical audience.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
sns.set_theme(style='whitegrid')
```

```python
df = px.data.gapminder()
df.head()
```

### Data validation
The Plotly documentation states that the Gapminder dataset contains one row per country and year and includes country, continent, year, life expectancy, population and GDP per capita.

```python
print('Rows:', len(df))
print('Columns:', df.columns.tolist())
print('Missing values:', df.isna().sum().sum())
print('Duplicate rows:', df.duplicated().sum())
```

## Visualization 1 — Life expectancy trend
The line chart establishes the first part of the story: health outcomes improved over time, but not at the same rate across continents.

```python
trend = df.groupby(['year','continent'], as_index=False)['lifeExp'].mean()
sns.lineplot(data=trend, x='year', y='lifeExp', hue='continent', marker='o')
plt.title('Life Expectancy Improved Across Every Continent')
plt.xlabel('Year'); plt.ylabel('Average Life Expectancy (years)')
plt.show()
```

## Visualization 2 — Income and life expectancy
A log-scaled scatter plot shows the broad relationship between GDP per capita and life expectancy in 2007. Bubble size adds population so the audience can see scale at the same time.

```python
latest = df[df.year == 2007].copy()
sns.scatterplot(data=latest, x='gdpPercap', y='lifeExp', size='pop', hue='continent', sizes=(30,900), alpha=.75)
plt.xscale('log')
plt.title('Higher Income Is Associated with Longer Life Expectancy (2007)')
plt.show()
```

## Visualization 3 — Population growth
The second line chart adds scale to the narrative and shows that population growth has been uneven across continents.

```python
popc = df.groupby(['year','continent'], as_index=False)['pop'].sum()
sns.lineplot(data=popc, x='year', y='pop', hue='continent', marker='o')
plt.title('Population Growth Has Been Uneven Across Continents')
plt.ylabel('Population')
plt.show()
```

## Visualization 4 — Distribution and inequality
The box-and-strip plot prevents averages from hiding variation within continents.

```python
sns.boxplot(data=latest, x='continent', y='lifeExp')
sns.stripplot(data=latest, x='continent', y='lifeExp', color='black', alpha=.35, size=3)
plt.title('Life Expectancy Still Varied Widely Within Continents (2007)')
plt.show()
```

## Visualization 5 — High-income outliers
A ranked bar chart makes the highest GDP-per-capita observations easy to compare and identify.

```python
top10 = latest.nlargest(10, 'gdpPercap').sort_values('gdpPercap')
sns.barplot(data=top10, y='country', x='gdpPercap', hue='continent', dodge=False, legend=False)
plt.title('Highest GDP per Capita Countries in 2007')
plt.show()
```

## Visualization 6 — Interactive map
The map turns a country-level table into a geographic story.

```python
fig = px.choropleth(latest, locations='iso_alpha', color='lifeExp', hover_name='country',
                    title='Global Life Expectancy by Country — 2007')
fig.show()
```

## Visualization 7 — Animated bubble chart
The animation connects the three major dimensions—income, health and population—across time and is useful for storytelling to a non-technical audience.

```python
fig = px.scatter(df, x='gdpPercap', y='lifeExp', size='pop', color='continent', hover_name='country',
                 animation_frame='year', log_x=True, size_max=55,
                 title='The Development Story: Income, Health and Population')
fig.show()
```
